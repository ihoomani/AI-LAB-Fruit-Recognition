# بررسی محیط اجرای پروژه تشخیص میوه

این نوت‌بوک برای ثبت اطلاعات محیط اجرای پروژه تهیه شده است. در این فایل، نسخه پایتون، مسیر مفسر، سیستم‌عامل و نسخه کتابخانه‌های اصلی بررسی می‌شوند.

**نکته:** سلول‌های این نوت‌بوک را در همان محیطی اجرا کنید که برای توسعه پروژه استفاده می‌کنید.

```python
import sys
import platform

print("نسخه پایتون:")
print(sys.version)

print("\nمسیر اجرای پایتون:")
print(sys.executable)

print("\nسیستم عامل و معماری:")
print(platform.platform())
```

## بررسی نسخه کتابخانه‌ها

در سلول بعدی، نسخه بسته‌های اصلی پروژه بررسی می‌شود. اگر بسته‌ای نصب نشده باشد، پیام «نصب نیست» نمایش داده خواهد شد.

```python
from importlib import metadata

packages = {
    "numpy": "NumPy",
    "matplotlib": "Matplotlib",
    "opencv-python": "OpenCV",
    "pillow": "Pillow",
    "scikit-learn": "scikit-learn"
}

for package_name, display_name in packages.items():
    try:
        version = metadata.version(package_name)
        print(f"{display_name}: {version}")
    except metadata.PackageNotFoundError:
        print(f"{display_name}: نصب نیست")
```

## بررسی وارد کردن کتابخانه‌های اصلی

این بخش بررسی می‌کند که NumPy، Matplotlib و OpenCV در محیط فعلی قابل وارد کردن باشند.

```python
modules = [
    ("numpy", "NumPy"),
    ("matplotlib", "Matplotlib"),
    ("cv2", "OpenCV")
]

for module_name, display_name in modules:
    try:
        module = __import__(module_name)
        version = getattr(module, "__version__", "نامشخص")
        print(f"{display_name}: نصب و قابل استفاده است (نسخه {version})")
    except ImportError:
        print(f"{display_name}: نصب نیست یا قابل وارد کردن نیست")
```
